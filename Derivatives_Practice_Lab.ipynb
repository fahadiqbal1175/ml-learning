## Informal definition of derivatives

The formal definition of derivatives can be a bit daunting with limits and values 'going to zero'. The idea is really much simpler.

The derivative of a function describes how the output of a function changes when there is a small change in an input variable.

Let's use a profit function $P(x)$ as an example, where $P$ is the output (profit) and $x$ is the input variable (units sold). Let's give a 'small change' a name *epsilon* or $\epsilon$. We use these Greek letters because it is traditional in mathematics to use *epsilon* ($\epsilon$) or *delta* ($\Delta$) to represent a small value. You can think of it as representing 0.001 or some other small value.

$$
\begin{equation}
\text{if } x \uparrow \epsilon \text{ causes }P(x) \uparrow \text{by }k \times \epsilon \text{ then}  \\
\frac{\partial P(x)}{\partial x} = k \tag{1}
\end{equation}
$$

This just says if you change the input to the function $P(x)$ by a little bit and the output changes by $k$ times that little bit, then the derivative of $P(x)$ is equal to $k$.

Let's try this out. Let's look at the derivative of the function $P(x) = x^2$ at the point $x=5$ and $\epsilon = 0.001$

````python
P = (5)**2
P_epsilon = (5 + 0.001)**2
k = (P_epsilon - P)/0.001    # difference divided by epsilon
print(f"P = {P}, P_epsilon = {P_epsilon}, dP_dx ~= k = {k:0.6f} ")
````

We increased the input value a little bit (0.001), causing the output to change from 25 to 25.010001, an increase of about 10 times the input increase. Referencing (1) above, this says that $k=10$, so $\frac{\partial P(x)}{\partial x} \approx 10$. If you are familiar with calculus, you know, written symbolically, $\frac{\partial P(x)}{\partial x} = 2x$. With $x=5$ this is 10. Our calculation above is not exactly 10 because to be exactly correct $\epsilon$ would need to be infinitesimally small. That is why we use the symbols $\approx$ or ~= rather than =. Let's see what happens if we make $\epsilon$ smaller.

````python
P = (5)**2
P_epsilon = (5 + 0.000000001)**2
k = (P_epsilon - P)/0.000000001
print(f"P = {P}, P_epsilon = {P_epsilon}, dP_dx ~= k = {k} ")
````

The value gets close to exactly 10 as we reduce the size of $\epsilon$. Feel free to try reducing the value further.

## Finding symbolic derivatives
In backprop it is useful to know the derivative of simple functions at any input value. Put another way, we would like to know the 'symbolic' derivative rather than the 'arithmetic' derivative. An example of a symbolic derivative is $\frac{\partial P(x)}{\partial x} = 2x$, the derivative of $P(x) = x^2$ above. With the symbolic derivative you can find the value of the derivative at any input value $x$.

If you have taken a calculus course, you are familiar with the many differentiation rules that mathematicians have developed to solve for a derivative given an expression. This process has been automated with symbolic differentiation programs. An example of this in python is the [SymPy](https://www.sympy.org/en/index.html) library. Let's take a look at how to use this.

### $y = x^2$
Define the python variables and their symbolic names.

````python
y, x = symbols('y, x')
````

Define and print the expression. Note SymPy produces a latex string which generates a nicely readable equation.

````python
y = x**2
y
````

Use SymPy's `diff` to differentiate the expression for $y$ with respect to $x$. Note the result matches our earlier example.

````python
dy_dx = diff(y,x)
dy_dx
````

Evaluate the derivative at a few points by 'substituting' numeric values for the symbolic values. In the first example, $x$ is replaced by $4$.

````python
dy_dx.subs([(x,4)])    # derivative at the point x = 4
````

````python
dy_dx.subs([(x,5)])    # derivative at the point x = 5
````

````python
dy_dx.subs([(x,-5)])    # derivative at the point x = -5
````

## $y = 4x$

````python
x, y = symbols('x, y')
````

````python
y = 4 * x
y
````

````python
dy_dx = diff(y,x)
dy_dx
````

````python
dy_dx.subs([(x,-6)])    # derivative at the point x = -6
````

Compare this with the arithmetic calculation

````python
y = 4*6
y_epsilon = 4*(6 + 0.001)
k = (y_epsilon - y)/0.001
print(f"y = {y}, y_epsilon = {y_epsilon}, dy_dx ~= k = {k} ")
````

For the function $y=4x$, it is easy to see that any change in $x$ will result in 4 times that amount of change in the output $y$, regardless of the starting value of $x$. Our arithmetic result confirms this.
